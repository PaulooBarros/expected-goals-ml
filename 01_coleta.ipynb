## 1. Importações

````python
import sys
from pathlib import Path

# Adiciona o diretório raiz ao path para importar módulos de src/
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from statsbombpy import sb

sns.set_style('whitegrid')
%matplotlib inline
````

## 2. Explorando as competições disponíveis

````python
competicoes = sb.competitions()
print(f'Total de competições disponíveis: {len(competicoes)}')
competicoes.head()
````

````python
# Filtrando as três competições do nosso escopo
torneios_alvo = [
    (43, 3),   # World Cup 2018
    (43, 106), # World Cup 2022
    (55, 43),  # Euro 2020
]

for comp_id, season_id in torneios_alvo:
    filtro = (competicoes['competition_id'] == comp_id) & (competicoes['season_id'] == season_id)
    print(competicoes[filtro][['competition_name', 'season_name', 'competition_gender']])
````

## 3. Coletando uma partida de exemplo

Vamos coletar uma partida da Copa 2022 para entender a estrutura dos dados.

````python
# Pegando as partidas da Copa 2022
partidas_wc22 = sb.matches(competition_id=43, season_id=106)
print(f'Total de partidas: {len(partidas_wc22)}')
partidas_wc22[['match_id', 'home_team', 'away_team', 'home_score', 'away_score']].head()
````

````python
# Coletando eventos de uma partida
match_id_exemplo = partidas_wc22.iloc[0]['match_id']
eventos = sb.events(match_id=match_id_exemplo)
print(f'Total de eventos: {len(eventos)}')
print(f'Tipos de eventos: {eventos["type"].unique()[:15]}')
````

## 4. Filtrando apenas finalizações (Shots)

````python
chutes = eventos[eventos['type'] == 'Shot'].copy()
print(f'Total de chutes na partida: {len(chutes)}')

# Colunas relevantes
colunas_relevantes = [
    'player', 'team', 'location',
    'shot_body_part', 'shot_type', 'shot_technique',
    'shot_outcome', 'shot_statsbomb_xg',
    'under_pressure', 'minute', 'period'
]
chutes[colunas_relevantes].head(10)
````
